# Tech Challenge Fase 1 — Análise do NPS em E-commerce

O objetivo deste projeto é analisar os fatores que impactam a satisfação do cliente em um e-commerce, medida por meio do Net Promoter Score (NPS), utilizando dados de pedidos, logística e atendimento ao cliente.

A análise busca transformar dados operacionais em insights acionáveis capazes de apoiar decisões estratégicas antes da aplicação da pesquisa de satisfação.

## Etapa 1 - Entendimento do Negócio

## Problema de Negócio

Com o crescimento acelerado do e-commerce, aumentou também o volume de pedidos, entregas e interações com clientes. Apesar disso, observa-se grande variabilidade na satisfação dos consumidores.

O principal desafio é entender quais fatores operacionais influenciam o NPS e como a empresa pode atuar preventivamente antes da aplicação da pesquisa de satisfação.

Atualmente, a empresa coleta o NPS apenas após a finalização da jornada do cliente, o que limita sua capacidade de atuação preventiva. Como consequência, as ações são tomadas de forma reativa, apenas após a ocorrência de experiências negativas.

A análise dos dados indica um alto volume de clientes detratores, evidenciando falhas relevantes na experiência do cliente. Nesse contexto, o principal problema de negócio consiste em identificar os fatores que levam à insatisfação e permitir que a empresa atue de forma antecipada.

## Importância do NPS

O NPS é uma métrica essencial para o e-commerce, pois está diretamente relacionado à satisfação, retenção e crescimento do negócio.

Clientes promotores tendem a recomprar e recomendar a marca, enquanto detratores podem impactar negativamente a reputação e reduzir a aquisição de novos clientes.

Áreas impactadas

- Logística: atrasos na entrega impactam diretamente a satisfação
- Atendimento: número de contatos e tempo de resolução afetam a experiência
- Operações: consistência da jornada do cliente
- Estratégia: tomada de decisão orientada a dados

## Relação entre NPS e Recompra

Os dados indicam que clientes com maior nível de satisfação apresentam maior probabilidade de recompra em 30 dias, reforçando o papel do NPS como indicador estratégico de fidelização.

## Impacto do NPS em Boca a Boca e Market Share

Clientes promotores contribuem para crescimento orgânico por meio de recomendação espontânea, enquanto detratores podem impactar negativamente a reputação da marca.

Em mercados competitivos como o e-commerce, a experiência do cliente influencia diretamente a participação de mercado.

## Indicadores que Complementam a Análise de NPS

Além do NPS, outros indicadores podem apoiar decisões estratégicas:

- SLA logístico
- Tempo médio de resolução
- Taxa de recompra
- Churn rate ou Taxa de cancelamento: indicando o número de clientes que cancelaram o contrato ou pararam de comprar, comparado ao total de clientes.
- Benchmark de NPS do setor
