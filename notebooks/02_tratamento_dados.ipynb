## ETAPA 2 - Definição da Target

# Qual variável representa a satisfação do cliente?

A variável que representa a satisfação do cliente neste projeto é o **NPS (Net Promoter Score)**, especificamente por meio da variável `nps_score`, que varia de 0 a 10.

Além disso, para facilitar a interpretação de negócio, o NPS também é analisado de forma categórica, classificando os clientes em:

- Detratores (0 a 6)
- Neutros (7 e 8)
- Promotores (9 e 10)


# Por que essa variável foi escolhida?

O NPS foi escolhido por ser uma métrica amplamente utilizada para medir a satisfação e lealdade do cliente.

Ele sintetiza a percepção geral da experiência em um único indicador, permitindo:

- Avaliar o nível de satisfação do cliente
- Identificar clientes com maior risco de churn (detratores)
- Medir o potencial de recomendação da marca (promotores)

Além disso, o NPS está diretamente relacionado a indicadores estratégicos do negócio, como retenção, recompra e crescimento.

# Em que momento da jornada essa informação é coletada?

O NPS é coletado **ao final da jornada do cliente**, após a conclusão da experiência de compra e entrega.

Isso significa que ele reflete uma avaliação retrospectiva, baseada na percepção completa da experiência, incluindo fatores como:

- Prazo de entrega
- Qualidade do atendimento
- Resolução de problemas

# Existe algum risco de usar essa variável de forma inadequada?

Sim, existem alguns riscos importantes que devem ser considerados:

- **Caráter reativo:** como o NPS é coletado após a experiência, ele não permite atuação preventiva direta
- **Viés de resposta:** clientes muito satisfeitos ou muito insatisfeitos tendem a responder mais, podendo distorcer a distribuição
- **Falta de granularidade:** o NPS resume a experiência em um único número, não explicando diretamente as causas da satisfação ou insatisfação
- **Uso isolado:** analisar o NPS sem considerar variáveis operacionais pode levar a conclusões superficiais

Dessa forma, o NPS deve ser utilizado em conjunto com outras variáveis para identificar os fatores que impactam a experiência do cliente.

# Conexão com o problema de negócio

Embora o NPS seja uma métrica coletada ao final da jornada, ele pode ser utilizado como variável alvo para entender os fatores que impactam a satisfação.

A partir dessa análise, é possível construir modelos e indicadores que permitam **antecipar a experiência do cliente**, transformando uma métrica reativa em uma ferramenta preditiva e estratégica.
